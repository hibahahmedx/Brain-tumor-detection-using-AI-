```python
from ultralytics import SAM
from ultralytics import YOLO

yolo_model = YOLO(r"C:\Users\Hiba\brain_tumor_detection\myenv\runs\detect\train\weights\best.pt")

results = yolo_model(r"C:\Users\Hiba\brain_tumor_detection\myenv\test images\meningioma.jpg")

sam_model = SAM("sam2_b.pt")

for result in results:
    class_ids = result.boxes.cls.int().tolist()
    if len(class_ids):
        boxes = result.boxes.xyxy
        sam_results = sam_model(result.orig_img, bboxes = boxes, verbose = False, save = True, device = 'cpu')
```
